How many bp are there in SLMQRT fastq? What is the expected coverage based on genome length for this? Development for another script we wanna use in the pipeline

```python
from Bio import SeqIO

path = "/groups/cress/projects/efactors/jack/insertion_mapping/data/SLMQRT/input/SLMQRT_5_5_dnaSES0086.fastq"
records = SeqIO.parse(path, "fastq")

sizes = []
for record in records:
    s = len(record.seq)
    sizes.append(s)
```

```python
total_bases = sum(sizes)
print(f"{total_bases}b read; ~{total_bases / 4e6}x coverage")
```
